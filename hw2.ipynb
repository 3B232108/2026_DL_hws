# HW2 批次學習

這次作業把 HW1 的全連接神經網路改成可以一次輸入 5 筆資料。

每筆資料有 6 個特徵值，網路結構和 HW1 相同。

## 1. 匯入套件與函數

Softmax 和 Cross Entropy Error 要改成可以處理多筆資料。

```python
import numpy as np

def relu(x):
    return np.maximum(0, x)

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def softmax(x):
    x = x - np.max(x, axis=1, keepdims=True)
    exp_x = np.exp(x)
    return exp_x / np.sum(exp_x, axis=1, keepdims=True)

def cross_entropy_error(y_hat, y):
    delta = 1e-7
    return -np.sum(y * np.log(y_hat + delta)) / y.shape[0]
```

## 2. 建立批次資料 X 和 Y

X 有 5 筆資料，每筆 6 個特徵值，所以形狀是 `(5, 6)`。

Y 是 5 筆資料的正確答案，使用 one-hot 格式，所以形狀是 `(5, 3)`。

```python
X = np.array([
    [0.5, 1.0, -0.5, 2.0, 0.3, 1.5],
    [1.0, 0.2, 0.8, -0.3, 1.2, 0.5],
    [0.1, 1.5, 0.4, 0.7, -0.2, 1.0],
    [1.2, -0.4, 0.3, 1.8, 0.6, 0.9],
    [0.7, 0.8, -0.1, 0.5, 1.4, 0.2]
])

Y = np.array([
    [0, 0, 1],
    [1, 0, 0],
    [0, 1, 0],
    [0, 0, 1],
    [0, 1, 0]
])

print("X =")
print(X)
print("X shape =", X.shape)

print("Y =")
print(Y)
print("Y shape =", Y.shape)
```

## 3. 權重與偏值

各層的權重 W 和偏值 b 與 HW1 相同。

```python
W1 = np.array([
    [0.2, -0.1, 0.4, 0.1, -0.3],
    [0.7, 0.2, -0.5, 0.3, 0.6],
    [-0.4, 0.5, 0.2, -0.2, 0.1],
    [0.3, -0.6, 0.1, 0.8, 0.2],
    [0.5, 0.4, -0.3, 0.2, -0.1],
    [-0.2, 0.3, 0.6, -0.4, 0.5]
])
b1 = np.array([0.1, -0.2, 0.05, 0.1, -0.05])

W2 = np.array([
    [0.3, -0.2, 0.4, 0.1],
    [-0.5, 0.6, 0.2, -0.1],
    [0.7, 0.1, -0.3, 0.5],
    [0.2, 0.4, 0.6, -0.2],
    [-0.1, 0.3, 0.2, 0.7]
])
b2 = np.array([0.1, 0.0, -0.1, 0.05])

W3 = np.array([
    [0.4, -0.3, 0.2, 0.5],
    [0.1, 0.6, -0.2, 0.3],
    [-0.5, 0.2, 0.7, -0.1],
    [0.3, 0.4, 0.1, 0.6]
])
b3 = np.array([0.0, 0.1, -0.05, 0.2])

W4 = np.array([
    [0.5, -0.2, 0.3],
    [-0.3, 0.6, 0.2],
    [0.4, 0.1, -0.5],
    [0.2, 0.3, 0.7]
])
b4 = np.array([0.1, -0.1, 0.05])

print("W1 shape:", W1.shape, "b1 shape:", b1.shape)
print("W2 shape:", W2.shape, "b2 shape:", b2.shape)
print("W3 shape:", W3.shape, "b3 shape:", b3.shape)
print("W4 shape:", W4.shape, "b4 shape:", b4.shape)
```

## 4. 第一隱藏層

5 筆資料一起和 W1 做矩陣運算，再經過 ReLU。

```python
z1 = np.dot(X, W1) + b1
h1 = relu(z1)

print("z1 =")
print(z1)
print("z1 shape =", z1.shape)

print("h1 =")
print(h1)
print("h1 shape =", h1.shape)
```

## 5. 第二隱藏層

第二層有 4 個神經元，使用 ReLU。

```python
z2 = np.dot(h1, W2) + b2
h2 = relu(z2)

print("z2 =")
print(z2)
print("z2 shape =", z2.shape)

print("h2 =")
print(h2)
print("h2 shape =", h2.shape)
```

## 6. 第三隱藏層

第三層有 4 個神經元，使用 Sigmoid。

```python
z3 = np.dot(h2, W3) + b3
h3 = sigmoid(z3)

print("z3 =")
print(z3)
print("z3 shape =", z3.shape)

print("h3 =")
print(h3)
print("h3 shape =", h3.shape)
```

## 7. 輸出層

輸出層有 3 個神經元，使用 Softmax。

每一列 y_hat 代表一筆資料對 3 個類別的預測機率。

```python
z4 = np.dot(h3, W4) + b4
y_hat = softmax(z4)

print("z4 =")
print(z4)
print("z4 shape =", z4.shape)

print("y_hat =")
print(y_hat)
print("y_hat shape =", y_hat.shape)
```

## 8. 計算平均 Cross Entropy Error

最後用 5 筆資料的 y_hat 和 Y 計算誤差，再取平均值。

```python
loss = cross_entropy_error(y_hat, Y)

print("Y =")
print(Y)

print("y_hat =")
print(y_hat)

print("平均 Cross Entropy Error =", loss)
```

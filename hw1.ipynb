## 1. 多層全連接神經網路

```python
import numpy as np

def relu(x):
    return np.maximum(0, x)

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def softmax(x):
    x = x - np.max(x)
    exp_x = np.exp(x)
    return exp_x / np.sum(exp_x)

def cross_entropy_error(y_hat, y):
    delta = 1e-7
    return -np.sum(y * np.log(y_hat + delta))
```

```python
x = np.array([0.5, 1.0, -0.5, 2.0, 0.3, 1.5])

W1 = np.array([
    [0.2, -0.1, 0.4, 0.1, -0.3],
    [0.7, 0.2, -0.5, 0.3, 0.6],
    [-0.4, 0.5, 0.2, -0.2, 0.1],
    [0.3, -0.6, 0.1, 0.8, 0.2],
    [0.5, 0.4, -0.3, 0.2, -0.1],
    [-0.2, 0.3, 0.6, -0.4, 0.5]
])
b1 = np.array([0.1, -0.2, 0.05, 0.1, -0.05])

W2 = np.array([
    [0.3, -0.2, 0.4, 0.1],
    [-0.5, 0.6, 0.2, -0.1],
    [0.7, 0.1, -0.3, 0.5],
    [0.2, 0.4, 0.6, -0.2],
    [-0.1, 0.3, 0.2, 0.7]
])
b2 = np.array([0.1, 0.0, -0.1, 0.05])

W3 = np.array([
    [0.4, -0.3, 0.2, 0.5],
    [0.1, 0.6, -0.2, 0.3],
    [-0.5, 0.2, 0.7, -0.1],
    [0.3, 0.4, 0.1, 0.6]
])
b3 = np.array([0.0, 0.1, -0.05, 0.2])

W4 = np.array([
    [0.5, -0.2, 0.3],
    [-0.3, 0.6, 0.2],
    [0.4, 0.1, -0.5],
    [0.2, 0.3, 0.7]
])
b4 = np.array([0.1, -0.1, 0.05])
```

```python
z1 = np.dot(x, W1) + b1
h1 = relu(z1)

print("W1 shape:", W1.shape, "b1 shape:", b1.shape)
print("W1參數:", W1.size, "b1參數:", b1.size)
print("z1:", z1)
print("z1 shape:", z1.shape)
print("h1:", h1)
print("h1 shape:", h1.shape)
```

```python
z2 = np.dot(h1, W2) + b2
h2 = relu(z2)

print("W2 shape:", W2.shape, "b2 shape:", b2.shape)
print("W2參數:", W2.size, "b2參數:", b2.size)
print("z2:", z2)
print("z2 shape:", z2.shape)
print("h2:", h2)
print("h2 shape:", h2.shape)
```

```python
z3 = np.dot(h2, W3) + b3
h3 = sigmoid(z3)

print("W3 shape:", W3.shape, "b3 shape:", b3.shape)
print("W3參數:", W3.size, "b3參數:", b3.size)
print("z3:", z3)
print("z3 shape:", z3.shape)
print("h3:", h3)
print("h3 shape:", h3.shape)
```

```python
z4 = np.dot(h3, W4) + b4
y_hat = softmax(z4)

print("W4 shape:", W4.shape, "b4 shape:", b4.shape)
print("W4參數:", W4.size, "b4參數:", b4.size)
print("z4:", z4)
print("z4 shape:", z4.shape)
print("y_hat:", y_hat)
print("y_hat shape:", y_hat.shape)
```

```python
total = W1.size + b1.size + W2.size + b2.size + W3.size + b3.size + W4.size + b4.size
print("總參數數量:", total)
```

```python
y = np.array([0, 0, 1])

loss = cross_entropy_error(y_hat, y)

print("y:", y)
print("y_hat:", y_hat)
print("cross entropy error:", loss)
```

## 2. dnn 函數

```python
def dnn(x):
    z1 = np.dot(x, W1) + b1
    h1 = relu(z1)

    z2 = np.dot(h1, W2) + b2
    h2 = relu(z2)

    z3 = np.dot(h2, W3) + b3
    h3 = sigmoid(z3)

    z4 = np.dot(h3, W4) + b4
    y_hat = softmax(z4)

    return y_hat
```

```python
result = dnn(x)

print("第一題 y_hat:", y_hat)
print("dnn y_hat:", result)
```

## 3. Gradio

```python
import gradio as gr

def predict(x1, x2, x3, x4, x5, x6):
    x = np.array([x1, x2, x3, x4, x5, x6], dtype=float)
    y_hat = dnn(x)
    result = np.argmax(y_hat) + 1
    return f"類別 {result}"

demo = gr.Interface(
    fn=predict,
    inputs=[
        gr.Number(label="x1"),
        gr.Number(label="x2"),
        gr.Number(label="x3"),
        gr.Number(label="x4"),
        gr.Number(label="x5"),
        gr.Number(label="x6")
    ],
    outputs=gr.Textbox(label="結果")
)

demo.launch()
```

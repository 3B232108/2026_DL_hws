# HW1 多層全連接神經網路

這次作業使用 NumPy 模擬一個多層全連接神經網路。

網路結構為：

6 個輸入 → 5 個神經元 → 4 個神經元 → 4 個神經元 → 3 個輸出

三個隱藏層分別使用 ReLU、ReLU、Sigmoid，輸出層使用 Softmax。

## 1. 匯入套件與函數

先匯入 NumPy，並建立這次會使用到的 ReLU、Sigmoid、Softmax 和 Cross Entropy Error。

```python
import numpy as np

def relu(x):
    return np.maximum(0, x)

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def softmax(x):
    x = x - np.max(x)
    exp_x = np.exp(x)
    return exp_x / np.sum(exp_x)

def cross_entropy_error(y_hat, y):
    delta = 1e-7
    return -np.sum(y * np.log(y_hat + delta))
```

## 2. 輸入資料與權重

輸入資料 x 有 6 個特徵值。

每一層的權重 W 和偏值 b 都自行設定固定數值。

```python
x = np.array([0.5, 1.0, -0.5, 2.0, 0.3, 1.5])

W1 = np.array([
    [0.2, -0.1, 0.4, 0.1, -0.3],
    [0.7, 0.2, -0.5, 0.3, 0.6],
    [-0.4, 0.5, 0.2, -0.2, 0.1],
    [0.3, -0.6, 0.1, 0.8, 0.2],
    [0.5, 0.4, -0.3, 0.2, -0.1],
    [-0.2, 0.3, 0.6, -0.4, 0.5]
])
b1 = np.array([0.1, -0.2, 0.05, 0.1, -0.05])

W2 = np.array([
    [0.3, -0.2, 0.4, 0.1],
    [-0.5, 0.6, 0.2, -0.1],
    [0.7, 0.1, -0.3, 0.5],
    [0.2, 0.4, 0.6, -0.2],
    [-0.1, 0.3, 0.2, 0.7]
])
b2 = np.array([0.1, 0.0, -0.1, 0.05])

W3 = np.array([
    [0.4, -0.3, 0.2, 0.5],
    [0.1, 0.6, -0.2, 0.3],
    [-0.5, 0.2, 0.7, -0.1],
    [0.3, 0.4, 0.1, 0.6]
])
b3 = np.array([0.0, 0.1, -0.05, 0.2])

W4 = np.array([
    [0.5, -0.2, 0.3],
    [-0.3, 0.6, 0.2],
    [0.4, 0.1, -0.5],
    [0.2, 0.3, 0.7]
])
b4 = np.array([0.1, -0.1, 0.05])

print("x =", x)
print("x shape =", x.shape)
```

## 3. 第一隱藏層

第一層有 5 個神經元，先計算 `z1 = xW1 + b1`，再使用 ReLU 得到 h1。

```python
z1 = np.dot(x, W1) + b1
h1 = relu(z1)

print("W1 shape:", W1.shape)
print("b1 shape:", b1.shape)
print("W1參數數量:", W1.size)
print("b1參數數量:", b1.size)

print("z1 =", z1)
print("z1 shape:", z1.shape)
print("h1 =", h1)
print("h1 shape:", h1.shape)
```

## 4. 第二隱藏層

第二層有 4 個神經元，同樣使用 ReLU。

```python
z2 = np.dot(h1, W2) + b2
h2 = relu(z2)

print("W2 shape:", W2.shape)
print("b2 shape:", b2.shape)
print("W2參數數量:", W2.size)
print("b2參數數量:", b2.size)

print("z2 =", z2)
print("z2 shape:", z2.shape)
print("h2 =", h2)
print("h2 shape:", h2.shape)
```

## 5. 第三隱藏層

第三層有 4 個神經元，這一層使用 Sigmoid。

```python
z3 = np.dot(h2, W3) + b3
h3 = sigmoid(z3)

print("W3 shape:", W3.shape)
print("b3 shape:", b3.shape)
print("W3參數數量:", W3.size)
print("b3參數數量:", b3.size)

print("z3 =", z3)
print("z3 shape:", z3.shape)
print("h3 =", h3)
print("h3 shape:", h3.shape)
```

## 6. 輸出層

輸出層有 3 個神經元，使用 Softmax 將結果轉成三個類別的機率。

y_hat 中的三個數值分別代表類別 1、類別 2、類別 3 的機率。

```python
z4 = np.dot(h3, W4) + b4
y_hat = softmax(z4)

print("W4 shape:", W4.shape)
print("b4 shape:", b4.shape)
print("W4參數數量:", W4.size)
print("b4參數數量:", b4.size)

print("z4 =", z4)
print("z4 shape:", z4.shape)
print("y_hat =", y_hat)
print("y_hat shape:", y_hat.shape)
```

## 7. 參數數量

每層參數數量就是 W 與 b 內所有數值的個數加總。

```python
total = (
    W1.size + b1.size +
    W2.size + b2.size +
    W3.size + b3.size +
    W4.size + b4.size
)

print("總參數數量 =", total)
```

## 8. Cross Entropy Error

這裡假設正確答案 y 為 `[0, 0, 1]`，代表正確類別是第 3 類。

接著利用 Cross Entropy Error 計算 y_hat 和 y 之間的誤差。

```python
y = np.array([0, 0, 1])

loss = cross_entropy_error(y_hat, y)

print("y =", y)
print("y_hat =", y_hat)
print("Cross Entropy Error =", loss)
```

# 第 2 題 dnn(x)

把前面的神經網路計算整理成一個 dnn(x) 函數。

權重和偏值都和第一題相同，所以輸入同一筆 x 時，結果應該和第一題的 y_hat 一樣。

```python
def dnn(x):
    z1 = np.dot(x, W1) + b1
    h1 = relu(z1)

    z2 = np.dot(h1, W2) + b2
    h2 = relu(z2)

    z3 = np.dot(h2, W3) + b3
    h3 = sigmoid(z3)

    z4 = np.dot(h3, W4) + b4
    y_hat = softmax(z4)

    return y_hat
```

```python
result = dnn(x)

print("第一題 y_hat =", y_hat)
print("dnn(x) =", result)
```

## 9. Gradio 互動介面

最後使用 Gradio 建立簡單介面，可以輸入 6 個特徵值。

資料會傳入 dnn(x)，再顯示機率最高的類別。

```python
import gradio as gr

def predict(x1, x2, x3, x4, x5, x6):
    x = np.array([x1, x2, x3, x4, x5, x6], dtype=float)
    y_hat = dnn(x)
    result = np.argmax(y_hat) + 1
    return f"類別 {result}"

demo = gr.Interface(
    fn=predict,
    inputs=[
        gr.Number(label="x1"),
        gr.Number(label="x2"),
        gr.Number(label="x3"),
        gr.Number(label="x4"),
        gr.Number(label="x5"),
        gr.Number(label="x6")
    ],
    outputs=gr.Textbox(label="結果")
)

demo.launch()
```
